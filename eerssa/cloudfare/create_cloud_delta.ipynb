# Creacion de Delta-Lake on R2 cloudfare

```python
import os
from deltalake import DeltaTable, write_deltalake
import pandas as pd

# Configure storage options
storage_options = {
    "AWS_ENDPOINT_URL": "https://<account>.r2.cloudflarestorage.com",
    "AWS_ACCESS_KEY_ID": "<​your-access-key>",
    "AWS_SECRET_ACCESS_KEY": "<​your-secret-key>",
    "AWS_REGION": "auto",
    "AWS_S3_LOCKING_PROVIDER": "dynamodb",  # needed for concurrent writes
}

# Read your local Delta table
local_table = DeltaTable("/path/to/your/delta-table")

# Write to R2
write_deltalake(
    "s3://my-deltalake-project/delta-table",
    local_table.to_pandas(),
    mode="overwrite",
    storage_options=storage_options
)
```
